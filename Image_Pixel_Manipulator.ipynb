```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
img = np.zeros((200,200,3), dtype=np.uint8)

print("Image Shape:", img.shape)
print("Image Data Type", img.dtype)
print("Image memory:", img.nbytes, "bytes")

plt.figure(figsize=(5,5))
plt.imshow(img)
plt.title("Original Blank Image")
plt.axis("Off")
plt.show()
```

```python
img[20:80, 20:80] = np.full((60,60,3), [255,0,0], dtype=np.uint8)
img[20:80, 80:140] = [0,255,0]

img[90:95,:] = [255,255,255]
img[:, 90:95] = [255,255,255]

plt.figure(figsize=(6,6))
plt.imshow(img)
plt.title("Synthetic Red Rectangle")
plt.axis("Off")
plt.show()
```
