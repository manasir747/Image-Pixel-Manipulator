```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
img = np.zeros((200,200,3), dtype=np.uint8)

print("Image Shape:", img.shape)
print("Image Data Type", img.dtype)
print("Image memory:", img.nbytes, "bytes")

plt.figure(figsize=(5,5))
plt.imshow(img)
plt.title("Original Blank Image")
plt.axis("Off")
plt.show()
```

```python
img[20:80, 20:80] = np.full((60,60,3), [255,0,0], dtype=np.uint8)
img[0:100, 0:100] = [255, 0, 0]
img[0:100, 100:200] = [0, 255, 0]
img[100:200, 0:100] = [0, 0, 255]
img[100:200, 100:200] = [255,255,0]

img[90:100,:] = [255,255,255]
img[:, 90:100] = [255,255,255]

plt.figure(figsize=(6,6))
plt.imshow(img)
plt.title("Synthetic Rectangles")
plt.axis("Off")
plt.show()
```

```python

# 1. Crop the image
cropped = img[50:150, 50:150, :]

# 2. Flip horizontally
horizontal_flip = img[:, ::-1, :]

# 3. Flip vertically
vertical_flip = img[::-1, :, :]

# Display all three side by side
plt.figure(figsize=(12, 4))

plt.subplot(1, 3, 1)
plt.imshow(cropped)
plt.title("Cropped Image")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(horizontal_flip)
plt.title("Horizontal Flip")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(vertical_flip)
plt.title("Vertical Flip")
plt.axis("off")

plt.tight_layout()
plt.show()



# 4. Extract individual RGB channels
red = img[:, :, 0]
green = img[:, :, 1]
blue = img[:, :, 2]

plt.figure(figsize=(12, 4))

plt.subplot(1, 3, 1)
plt.imshow(red, cmap="gray")
plt.title("Red Channel")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(green, cmap="gray")
plt.title("Green Channel")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(blue, cmap="gray")
plt.title("Blue Channel")
plt.axis("off")

plt.show()
```

```python
img_float = img.astype(np.float32)

bright_img = np.clip(img_float + 50, 0, 255).astype(np.uint8)
dark_img = np.clip(img_float - 50, 0, 255).astype(np.uint8)
mask = np.all(img >= 200, axis=2)

plt.figure(figsize=(12,4))

plt.subplot(1,3,1)
plt.imshow(bright_img)
plt.title("Bright Image")
plt.axis("Off")

plt.subplot(1,3,2)
plt.imshow(dark_img)
plt.title("Darkened Image")
plt.axis("Off")

plt.subplot(1,3,3)
plt.imshow(mask, cmap='gray')
plt.title("Bright Pixels Mask")
plt.axis("Off")
```

```python
img_uint8 = img.astype(np.uint8)
img_float64 = img.astype(np.float64)

print("uint8 image memory:", img_uint8.nbytes, "bytes")
print("float64 image memory:", img_float64.nbytes, "bytes")

plt.figure(figsize=(10,4))

plt.subplot(1,2,1)
plt.imshow(img_uint8)
plt.title("uint8 image")
plt.axis("Off")
plt.subplot(1,2,2)
plt.imshow(img_float64)
plt.title("float64 image")
plt.axis("Off")
```
